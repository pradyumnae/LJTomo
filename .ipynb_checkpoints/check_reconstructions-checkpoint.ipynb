# LeJEPA v2 Reconstruction Quality Check
This notebook displays samples from the 2,677 reconstructed volumes.

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
import os

data_dir = 'data/lejepa_v2/'
samples = [
    'recon_20260305_100243_N55drp.h5',  # Lisabeth
    'recon_20251029_014438_RTV_MB_7A_x00y02.h5',  # RTV (Inverted)
    'recon_20230925_154748_testScan1.h5'  # Test Scan
]

for f_name in samples:
    path = os.path.join(data_dir, f_name)
    if not os.path.exists(path):
        continue
        
    with h5py.File(path, 'r') as f:
        # Load a middle slice
        data = f['reconstruction']
        mid = data.shape[0] // 2
        slice_data = data[mid]
        
        plt.figure(figsize=(12, 10))
        # Use percentile clipping for robust display
        vmin, vmax = np.percentile(slice_data, [1, 99])
        img = plt.imshow(slice_data, cmap='gray', vmin=vmin, vmax=vmax)
        plt.title(f"{f_name} (Mid Slice)")
        plt.colorbar(img, label='Absorption')
        plt.show()
        
        # Histogram to see the distribution
        plt.figure(figsize=(10, 4))
        plt.hist(slice_data.flatten(), bins=100)
        plt.title(f"Histogram: {f_name}")
        plt.axvline(vmin, color='r', linestyle='--', label='1st Pct')
        plt.axvline(vmax, color='g', linestyle='--', label='99th Pct')
        plt.legend()
        plt.show()
```
